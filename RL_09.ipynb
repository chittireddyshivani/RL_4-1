```python

# Assignment 1: Bellman Expectation Equations

gamma = 0.9

# Solve state values algebraically
V1 = 89.81 / 0.9271
V0 = -1 + 0.81 * V1

# Calculate action values
Q_S0_Down = -1 + gamma * V1
Q_S0_Right = -1
Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V0

print("ASSIGNMENT 1 RESULTS")
print("--------------------")
print(f"V_pi(S0)       = {V0:.2f}")
print(f"V_pi(S1)       = {V1:.2f}")
print(f"Q_pi(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q_pi(S0, Right)= {Q_S0_Right:.2f}")
print(f"Q_pi(S1, Right)= {Q_S1_Right:.2f}")
print(f"Q_pi(S1, Up)   = {Q_S1_Up:.2f}")

# Verify Bellman expectation equations
check_V0 = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
check_V1 = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print("\nVERIFICATION")
print(f"Verified V(S0) = {check_V0:.2f}")
print(f"Verified V(S1) = {check_V1:.2f}")
print("Verification successful:",
      abs(V0 - check_V0) < 1e-9
      and abs(V1 - check_V1) < 1e-9)
```

```python

# Assignment 2: Iterative Policy Evaluation
# Reinforcement Learning - 23CS201SE403

gamma = 0.9
theta = 0.000001

# Initialize state values
V = {
    "S0": 0.0,
    "S1": 0.0
}

# Fixed deterministic policy
policy = {
    "S0": "Down",
    "S1": "Right"
}

# Stochastic transitions:
# (next state, probability, reward)
transitions = {
    ("S0", "Down"): [
        ("Wall", 0.1, -1),
        ("S1", 0.9, -1)
    ],
    ("S1", "Right"): [
        ("G", 0.9, 100),
        ("S0", 0.1, -1)
    ],
    ("S0", "Right"): [
        ("Wall", 1.0, -1)
    ],
    ("S1", "Up"): [
        ("S0", 1.0, -1)
    ]
}

# Iterative policy evaluation
iterations = 0

while True:
    old_V = V.copy()
    new_V = {}

    for state in ["S0", "S1"]:
        action = policy[state]
        value = 0.0

        for next_state, probability, reward in transitions[
            (state, action)
        ]:
            next_value = old_V.get(next_state, 0.0)
            value += probability * (
                reward + gamma * next_value
            )

        new_V[state] = value

    delta = max(
        abs(new_V["S0"] - old_V["S0"]),
        abs(new_V["S1"] - old_V["S1"])
    )

    V = new_V
    iterations += 1

    if delta < theta:
        break

# Calculate action values
Q = {}

for (state, action), outcomes in transitions.items():
    value = 0.0

    for next_state, probability, reward in outcomes:
        next_value = V.get(next_state, 0.0)
        value += probability * (
            reward + gamma * next_value
        )

    Q[(state, action)] = value

# Display results
print("ASSIGNMENT 2 RESULTS")
print("--------------------")
print("Number of iterations:", iterations)
print(f"V_pi(S0) = {V['S0']:.2f}")
print(f"V_pi(S1) = {V['S1']:.2f}")

print("\nACTION VALUES")
print(f"Q_pi(S0, Down)  = {Q[('S0', 'Down')]:.2f}")
print(f"Q_pi(S0, Right) = {Q[('S0', 'Right')]:.2f}")
print(f"Q_pi(S1, Right) = {Q[('S1', 'Right')]:.2f}")
print(f"Q_pi(S1, Up)    = {Q[('S1', 'Up')]:.2f}")

# Verify policy consistency
print("\nPOLICY CONSISTENCY CHECKS")
print(
    "V(S0) == Q(S0, Down):",
    abs(V["S0"] - Q[("S0", "Down")]) < theta
)
print(
    "V(S1) == Q(S1, Right):",
    abs(V["S1"] - Q[("S1", "Right")]) < theta
)
```
