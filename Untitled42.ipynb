```python
# ============================================================
# ASSIGNMENT 1
# Calculate Undiscounted and Discounted Returns
# ============================================================

# Accept number of time steps
n = int(input("Enter number of time steps/episodes: "))

# Accept reward for each time step
rewards = []

for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)

# Calculate Undiscounted Return
undiscounted_return = sum(rewards)

# Accept discount factor
gamma = float(input("Enter discount factor (gamma): "))

# Calculate Discounted Return
discounted_return = 0

for t in range(n):
    discounted_return += (gamma ** t) * rewards[t]

# Display results
print("\n======================================")
print("        RETURN CALCULATION")
print("======================================")

print("Entered reward values:", rewards)

print("Undiscounted Return:", undiscounted_return)

print("Discounted Return:", discounted_return)

print("======================================")
```

```python
# ============================================================
# Create grid_env_det.py
# ============================================================

%%writefile grid_env_det.py

class GridWorldMDP:

    def __init__(self):

        # ----------------------------------------------------
        # 1. STATE SPACE
        # ----------------------------------------------------
        self.states = [
            "S0",
            "S1",
            "S3",
            "S4",
            "G"
        ]

        # ----------------------------------------------------
        # 2. ACTION SPACE
        # ----------------------------------------------------
        self.actions = [
            "up",
            "down",
            "left",
            "right"
        ]

        # ----------------------------------------------------
        # 3. REWARD FUNCTION
        # ----------------------------------------------------
        self.normal_reward = -1
        self.goal_reward = 10
        self.obstacle_reward = -5

        # ----------------------------------------------------
        # 4. DISCOUNT FACTOR
        # ----------------------------------------------------
        self.gamma = 0.9

        # ----------------------------------------------------
        # 5. TRANSITION FUNCTION
        # ----------------------------------------------------
        self.transitions = {

            "S0": {
                "up": "S0",
                "down": "S0",
                "left": "S0",
                "right": "S1"
            },

            "S1": {
                "up": "S1",
                "down": "S3",
                "left": "S0",
                "right": "G"
            },

            "S3": {
                "up": "S1",
                "down": "S3",
                "left": "S3",
                "right": "S4"
            },

            "S4": {
                "up": "G",
                "down": "S4",
                "left": "S3",
                "right": "S4"
            },

            "G": {
                "up": "G",
                "down": "G",
                "left": "G",
                "right": "G"
            }
        }

    # --------------------------------------------------------
    # Transition Function
    # --------------------------------------------------------
    def transition(self, state, action):

        if state not in self.states:
            raise ValueError("Invalid state!")

        if action not in self.actions:
            raise ValueError("Invalid action!")

        return self.transitions[state][action]

    # --------------------------------------------------------
    # Reward Function
    # --------------------------------------------------------
    def reward(self, state, action):

        next_state = self.transition(state, action)

        if next_state == "G":
            return self.goal_reward

        return self.normal_reward

    # --------------------------------------------------------
    # Step Function
    # --------------------------------------------------------
    def step(self, state, action):

        next_state = self.transition(state, action)

        reward = self.reward(state, action)

        done = (next_state == "G")

        return next_state, reward, done
```

```python
# ============================================================
# ASSIGNMENT 2
# Grid World MDP
# ============================================================

from grid_env_det import GridWorldMDP

# Create environment
env = GridWorldMDP()

# ------------------------------------------------------------
# Display State Space
# ------------------------------------------------------------

print("State Space:")
print(env.states)

# ------------------------------------------------------------
# Display Action Space
# ------------------------------------------------------------

print("\nAction Space:")
print(env.actions)

# ------------------------------------------------------------
# Display Rewards
# ------------------------------------------------------------

print("\nRewards:")
print("Normal State Reward:", env.normal_reward)
print("Goal Reward:", env.goal_reward)
print("Obstacle Reward:", env.obstacle_reward)

# ------------------------------------------------------------
# Display Discount Factor
# ------------------------------------------------------------

print("\nDiscount Factor:")
print(env.gamma)

# ------------------------------------------------------------
# Display Transition Function
# ------------------------------------------------------------

print("\nTransition Function:")

for state in env.states:

    print("\nState:", state)

    for action in env.actions:

        next_state = env.transition(state, action)

        print(action, "->", next_state)
```
