```python
# Bellman Optimality and Value Iteration in a Deterministic 2x2 Grid World
# Layout:
#   S0  S1
#   WALL G
# Actions that hit a boundary or wall leave the agent in its current state and give -1.
# Reaching G gives +10; G is terminal and V(G)=0.

GAMMA = 0.9
THETA = 1e-6
NORMAL_REWARD = -1
WALL_REWARD = -1
GOAL_REWARD = 10

states = ["S0", "S1", "G"]
non_terminal_states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]
terminal_state = "G"

# transition[state][action] = (next_state, immediate_reward)
transition = {
    "S0": {
        "Up":    ("S0", WALL_REWARD),  # boundary
        "Down":  ("S0", WALL_REWARD),  # wall
        "Left":  ("S0", WALL_REWARD),  # boundary
        "Right": ("S1", NORMAL_REWARD)
    },
    "S1": {
        "Up":    ("S1", WALL_REWARD),  # boundary
        "Down":  ("G", GOAL_REWARD),
        "Left":  ("S0", NORMAL_REWARD),
        "Right": ("S1", WALL_REWARD)   # boundary
    }
}

print("1. MDP DETAILS")
print("States:", states)
print("Non-terminal states:", non_terminal_states)
print("Actions:", actions)
print(f"Discount factor gamma = {GAMMA}")
print(f"Rewards: normal={NORMAL_REWARD}, wall/boundary={WALL_REWARD}, goal={GOAL_REWARD}")
print("Grid layout: S0 | S1")
print("            WALL | G")
print("\\n2. TRANSITION VERIFICATION")
for state in non_terminal_states:
    for action in actions:
        next_state, reward = transition[state][action]
        print(f"{state:>2} --{action:<5}--> {next_state:<2} | reward = {reward}")
        assert next_state in states
    assert set(transition[state].keys()) == set(actions)
print("All non-terminal state-action pairs have defined transitions.")

# Synchronous value iteration: every update uses only values from the previous iteration.
V = {state: 0.0 for state in states}
V[terminal_state] = 0.0
print("\\n3. INITIAL VALUES")
print(V)

print("\\n4-5. VALUE ITERATION TRACE")
print(f"{'Iteration':>9} {'V(S0)':>14} {'V(S1)':>14} {'Delta':>14}")
iteration = 0
print(f"{iteration:>9} {V['S0']:>14.6f} {V['S1']:>14.6f} {'-':>14}")
while True:
    iteration += 1
    old_V = V.copy()
    new_V = old_V.copy()
    for state in non_terminal_states:
        action_values = []
        for action in actions:
            next_state, reward = transition[state][action]
            action_values.append(reward + GAMMA * old_V[next_state])
        new_V[state] = max(action_values)
    delta = max(abs(new_V[s] - old_V[s]) for s in non_terminal_states)
    V = new_V
    print(f"{iteration:>9} {V['S0']:>14.6f} {V['S1']:>14.6f} {delta:>14.8f}")
    if delta < THETA:
        break

print("\\n6. FINAL OPTIMAL STATE VALUES")
for state in states:
    print(f"V*({state}) = {V[state]:.6f}")

print("\\n7. ACTION VALUES Q*(s,a)")
Q = {}
for state in non_terminal_states:
    Q[state] = {}
    for action in actions:
        next_state, reward = transition[state][action]
        Q[state][action] = reward + GAMMA * V[next_state]
        print(f"Q*({state}, {action}) = {Q[state][action]:.6f}")

print("\\n8. OPTIMAL POLICY (all maximizing actions)")
policy = {}
for state in non_terminal_states:
    best_value = max(Q[state].values())
    best_actions = [a for a, q in Q[state].items() if abs(q - best_value) < 1e-9]
    policy[state] = best_actions
    print(f"pi*({state}) = {best_actions} (value = {best_value:.6f})")

print("\\n9. BELLMAN OPTIMALITY VERIFICATION")
for state in non_terminal_states:
    max_q = max(Q[state].values())
    print(f"{state}: V* = {V[state]:.6f}, max_a Q* = {max_q:.6f}, match = {abs(V[state]-max_q) < 1e-9}")
    assert abs(V[state] - max_q) < 1e-9

print("\\nEXPLANATION")
print("- From S1, Down reaches the goal and earns +10 immediately, so it is optimal.")
print("- From S0, Right moves to S1 for -1, then the next optimal action reaches G for +10.")
print("- gamma=0.9 discounts future rewards; the reward one step later contributes 0.9 times its value.")
print("- Synchronous updates use a separate old-value snapshot, avoiding in-place update order effects.")
```
